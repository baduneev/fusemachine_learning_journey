### Q3 — Answer the VP's question: P(θ_A > θ_B)

Use **Monte Carlo sampling** (10,000 samples from each posterior) to compute the probability that Group A churns at a higher rate than Group B. No p-value permitted.


````python
# Number of Monte Carlo samples to draw from each posterior distribution
# More samples = more stable probability estimate
MC_SAMPLES = 10_000

# Set random seed so that results are reproducible
# This means every time we run the cell, we get the same random samples
np.random.seed(42)

# ------------------------------------------------------------
# Sample possible churn probabilities from Group A posterior
# ------------------------------------------------------------

# Group A posterior is:
# Beta(alpha_prior + k_A, beta_prior + (n_A - k_A))
#
# k_A = number of churned customers in Group A
# n_A - k_A = number of non-churned customers in Group A
#
# np.random.beta(...) draws random churn probability values from this posterior
post_A_samples = np.random.beta(
    alpha_prior + k_A,          # posterior alpha for Group A
    beta_prior + (n_A - k_A),   # posterior beta for Group A
    MC_SAMPLES                  # number of random samples to draw
)

# ------------------------------------------------------------
# Sample possible churn probabilities from Group B posterior
# ------------------------------------------------------------

# Group B posterior is:
# Beta(alpha_prior + k_B, beta_prior + (n_B - k_B))
#
# k_B = number of churned customers in Group B
# n_B - k_B = number of non-churned customers in Group B
post_B_samples = np.random.beta(
    alpha_prior + k_B,          # posterior alpha for Group B
    beta_prior + (n_B - k_B),   # posterior beta for Group B
    MC_SAMPLES                  # number of random samples to draw
)

# ------------------------------------------------------------
# Compare Group A and Group B sample-by-sample
# ------------------------------------------------------------

# post_A_samples > post_B_samples returns True whenever
# sampled churn probability of Group A is greater than Group B
#
# Example:
# A sample = 0.42, B sample = 0.03 → True
# A sample = 0.25, B sample = 0.30 → False
#
# np.mean(True/False values) gives the fraction of times this is true
# This fraction estimates P(θ_A > θ_B)
p_A_greater = np.mean(post_A_samples > post_B_samples)

# ------------------------------------------------------------
# Self-check
# ------------------------------------------------------------

# This checks whether the probability is greater than 0.90
# The assignment expects Group A to churn at a higher rate than Group B
# with very high probability
assert 0.90 < p_A_greater <= 1.0, f"Expected P > 0.90, got {p_A_greater:.4f}"

# Print the final probability
print(f"P(θ_A > θ_B) = {p_A_greater:.4f}")

# Print business-friendly interpretation
# This answers the VP's question without using a p-value
print(
    f"We can tell the VP that Month-to-month churns at a higher rate with "
    f"{p_A_greater*100:.1f}% probability — without a p-value."
)
````

### Q4 — Implement the sequential update function


````python
def update_posterior(alpha, beta, churn_label):
    """
    Update Beta(alpha, beta) posterior after observing one customer.

    In this assignment:
    - churn_label = 1 means the customer churned
    - churn_label = 0 means the customer stayed

    Beta distribution has two parameters:
    - alpha represents churn/success count
    - beta represents non-churn/failure count

    So:
    - if customer churned, alpha increases by 1
    - if customer did not churn, beta increases by 1
    """

    # If the observed customer churned
    if churn_label == 1:

        # Increase alpha by 1 because churn is treated as a "success"
        alpha_new = alpha + 1

        # Beta stays the same because this was not a non-churn event
        beta_new = beta

    # If the observed customer did not churn
    else:

        # Alpha stays the same because this was not a churn event
        alpha_new = alpha

        # Increase beta by 1 because non-churn is treated as a "failure"
        beta_new = beta + 1

    # Return the updated posterior parameters
    return alpha_new, beta_new


# ── SELF-CHECK ────────────────────────────────────────────────────────────────

# Initial prior/posterior parameters
# Beta(2, 8) means:
# alpha = 2 pseudo-churn observations
# beta = 8 pseudo-non-churn observations
a0, b0 = 2.0, 8.0

# Case 1: observe one churn event
# Since churn_label = 1, alpha should increase by 1
a1, b1 = update_posterior(a0, b0, 1)

# Case 2: observe one non-churn event
# Since churn_label = 0, beta should increase by 1
a2, b2 = update_posterior(a0, b0, 0)

# Check that after churn, alpha increased and beta stayed same
assert a1 == a0 + 1 and b1 == b0, "After churn: alpha should increase by 1"

# Check that after no-churn, alpha stayed same and beta increased
assert a2 == a0 and b2 == b0 + 1, "After no-churn: beta should increase by 1"

# If both checks pass, function is correct
print("✅ update_posterior() is correct!")
````

### Q5 — Run sequential update and plot posterior evolution


````python
# Shuffle the dataset so that customer rows appear in random order.
# frac=1 means use 100% of rows but shuffled.
# random_state=42 makes the shuffle reproducible.
# reset_index(drop=True) resets row numbers after shuffling.
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)

# TRUE_RATE is the actual churn rate in the full dataset.
# Since Churn is encoded as 1 = churned, 0 = stayed,
# df["Churn"].mean() gives the proportion of churned customers.
TRUE_RATE = df["Churn"].mean()

# These are the sample sizes where we want to take snapshots.
# Example: after seeing 1 customer, 5 customers, 20 customers, etc.
snapshots = [1, 5, 20, 80, 200, 500]

# Initial prior belief: Beta(2, 8)
# alpha_init = prior churn count / success count
# beta_init  = prior non-churn count / failure count
alpha_init, beta_init = 2.0, 8.0

# This dictionary will store posterior parameters at each snapshot.
# Format:
# history[n] = (alpha_after_n_customers, beta_after_n_customers)
history = {}

# Start sequential updating from the prior.
alpha_current = alpha_init
beta_current = beta_init

# Loop through the first 500 rows of the shuffled dataset.
# i is row number starting from 1.
# churn_label is the observed churn value for that customer.
for i, churn_label in enumerate(df_shuffled["Churn"].iloc[:500], start=1):
    
    # Update posterior using one observation.
    # If churn_label = 1, alpha increases by 1.
    # If churn_label = 0, beta increases by 1.
    alpha_current, beta_current = update_posterior(
        alpha_current,
        beta_current,
        churn_label
    )
    
    # If current observation count is one of the snapshot points,
    # save the current posterior parameters.
    if i in snapshots:
        history[i] = (alpha_current, beta_current)


# Create possible churn probability values from 0 to 1.
# These values are used on the x-axis for plotting Beta PDFs.
theta_r = np.linspace(0.001, 0.999, 500)

# Create 2 rows and 3 columns of plots.
# Total = 6 plots, one for each snapshot.
fig, axes = plt.subplots(2, 3, figsize=(18, 8))

# Loop through each subplot and each snapshot value.
# axes.flat converts the 2x3 axes grid into a simple list of 6 axes.
for ax, n in zip(axes.flat, snapshots):
    
    # Get posterior alpha and beta saved at snapshot n.
    a_n, b_n = history[n]
    
    # Compute the Beta posterior density for this snapshot.
    # This shows our belief about churn probability after seeing n customers.
    pdf = beta_dist.pdf(theta_r, a_n, b_n)
    
    # Compute posterior mode.
    # This is the most likely churn probability under Beta(a_n, b_n).
    mode = (a_n - 1) / (a_n + b_n - 2)
    
    # Plot the posterior PDF curve.
    ax.plot(theta_r, pdf, color="steelblue", lw=2.5)
    
    # Plot vertical dashed red line for true full-dataset churn rate.
    # This is used as a reference line.
    ax.axvline(
        TRUE_RATE,
        color="crimson",
        ls="--",
        lw=2,
        label=f"True rate={TRUE_RATE:.3f}"
    )
    
    # Plot vertical orange dash-dot line for posterior mode.
    # This shows the current Bayesian estimate after n observations.
    ax.axvline(
        mode,
        color="orange",
        ls="-.",
        lw=2,
        label=f"Post mode={mode:.3f}"
    )
    
    # Set title showing number of observed customers and posterior parameters.
    ax.set_title(f"n={n}  Beta({a_n:.0f},{b_n:.0f})", fontsize=11)
    
    # Label x-axis as churn probability.
    ax.set_xlabel("θ")
    
    # Label y-axis as density.
    ax.set_ylabel("Density")
    
    # Show legend for true rate and posterior mode.
    ax.legend(fontsize=8)

# Add one main title for the whole figure.
plt.suptitle("Sequential Bayesian Updating: Beta-Binomial", fontsize=13)

# Adjust plot spacing so labels and titles do not overlap.
plt.tight_layout()

# Display the final plot.
plt.show()
````

### Q6 — Decision boundary: P(θ > 0.25) vs n


````python
# ------------------------------------------------------------
# Bayesian decision boundary:
# We want to track P(θ > 0.25) as more data arrives.
#
# θ = true churn probability
# threshold = 0.25 means we are asking:
# "What is the probability that the true churn rate is greater than 25%?"
# ------------------------------------------------------------

# Decision threshold for churn probability
# We want to know whether true churn rate is probably above 25%
threshold = 0.25

# Number of Monte Carlo samples drawn from the posterior at each step
# More samples give a more stable probability estimate
MC_SAMPLES = 10_000

# Set random seed for reproducibility
# This makes the random sampling result repeatable
np.random.seed(0)

# This list will store P(θ > 0.25) after each observation
# It will contain 500 values, one value for each n from 1 to 500
p_exceed = []

# Start with the prior Beta(2, 8)
# alpha_seq represents churn evidence
# beta_seq represents non-churn evidence
alpha_seq, beta_seq = 2.0, 8.0

# ------------------------------------------------------------
# Sequential Bayesian updating
# ------------------------------------------------------------

# Loop through the first 500 shuffled customers
# enumerate(..., start=1) gives n = 1, 2, 3, ..., 500
# churn_label is 1 if customer churned, 0 if customer stayed
for n, churn_label in enumerate(df_shuffled["Churn"].iloc[:500], start=1):

    # Update posterior after seeing one customer's churn outcome
    # If churn_label = 1, alpha increases
    # If churn_label = 0, beta increases
    alpha_seq, beta_seq = update_posterior(
        alpha_seq,
        beta_seq,
        churn_label
    )

    # Draw 10,000 possible churn-rate values from the current posterior
    # Current posterior is Beta(alpha_seq, beta_seq)
    posterior_samples = np.random.beta(
        alpha_seq,
        beta_seq,
        MC_SAMPLES
    )

    # Compute probability that θ is greater than the threshold 0.25
    # posterior_samples > threshold gives True/False values
    # np.mean(...) gives fraction of True values
    p_theta_above_threshold = np.mean(posterior_samples > threshold)

    # Store this probability for the current sample size n
    p_exceed.append(p_theta_above_threshold)


# ------------------------------------------------------------
# Frequentist sample size calculation
# ------------------------------------------------------------

# Import normal distribution functions
# norm.ppf gives z-scores for required confidence/power levels
from scipy.stats import norm

# z_a is the two-sided 5% significance critical value
# norm.ppf(0.975) ≈ 1.96
z_a = norm.ppf(0.975)

# z_b is the z-score for 80% power
# norm.ppf(0.80) ≈ 0.84
z_b = norm.ppf(0.80)

# p0 is the null/reference churn rate
# Here p0 = 0.25
p0 = threshold

# p1 is the expected/actual churn rate from the full dataset
# TRUE_RATE was computed earlier as df["Churn"].mean()
p1 = TRUE_RATE

# Approximate frequentist sample size for one-proportion z-test
# This asks: how many samples are needed to detect difference between p0 and p1
# with alpha=0.05 and power=0.80?
freq_n = int(
    np.ceil(
        (
            z_a * np.sqrt(p0 * (1 - p0)) +
            z_b * np.sqrt(p1 * (1 - p1))
        ) ** 2
        / (p1 - p0) ** 2
    )
)


# ------------------------------------------------------------
# Bayesian threshold crossing
# ------------------------------------------------------------

# Find first n where Bayesian probability P(θ > 0.25) becomes greater than 0.90
# enumerate(p_exceed, start=1) gives n and probability at that n
# next(...) returns first matching n
# If no value crosses 0.90, bayes_n becomes None
bayes_n = next(
    (n for n, p in enumerate(p_exceed, start=1) if p > 0.90),
    None
)


# ------------------------------------------------------------
# Plot Bayesian probability curve and compare with frequentist n
# ------------------------------------------------------------

# Create figure and axis
fig, ax = plt.subplots(figsize=(12, 5))

# Plot P(θ > 0.25) as sample size increases
ax.plot(
    range(1, len(p_exceed) + 1),
    p_exceed,
    color="steelblue",
    lw=2.5
)

# Draw horizontal line at 0.90
# This is the Bayesian decision threshold
ax.axhline(
    0.90,
    color="crimson",
    ls="--",
    lw=2,
    label="90% threshold"
)

# If Bayesian probability crosses 0.90,
# draw a vertical line at the first crossing point
if bayes_n:
    ax.axvline(
        bayes_n,
        color="steelblue",
        ls=":",
        lw=2,
        label=f"Bayesian: n={bayes_n}"
    )

# Draw vertical line for frequentist required sample size
ax.axvline(
    freq_n,
    color="orange",
    ls="-.",
    lw=2,
    label=f"Frequentist: n={freq_n}"
)

# Label x-axis
ax.set_xlabel("n")

# Label y-axis
ax.set_ylabel("P(θ > 0.25)")

# Add legend
ax.legend()

# Add title
ax.set_title("Bayesian Decision Boundary vs Frequentist Sample Size")

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()

# Print final comparison
print(f"Bayesian: n = {bayes_n}  |  Frequentist: n = {freq_n}")
````

### Q7 — Dirichlet-Multinomial: 3-category contract type


````python
# ------------------------------------------------------------
# Dirichlet-Multinomial model for contract category proportions
# ------------------------------------------------------------
# Goal:
# We want to estimate the proportion of customers in each contract category:
# 1. Month-to-month
# 2. One year
# 3. Two year
#
# Since there are more than two categories, we use a Dirichlet prior.
# Dirichlet is the multi-class version of Beta.
# Beta = for binary probability
# Dirichlet = for multiple category probabilities
# ------------------------------------------------------------


# These are the three possible contract categories in the Telco dataset
categories = ["Month-to-month", "One year", "Two year"]

# Flat Dirichlet prior
# [1, 1, 1] means before seeing data, we treat all categories equally likely.
# It adds 1 pseudo-count to each category.
prior_alpha = np.array([1.0, 1.0, 1.0])


# ------------------------------------------------------------
# 1. Count observations for each contract category
# ------------------------------------------------------------

# Count how many customers belong to each category.
# reindex(categories) ensures the counts are in the same order as our categories list.
# .values converts the pandas Series into a NumPy array.
counts = df["Contract"].value_counts().reindex(categories).values


# ------------------------------------------------------------
# 2. Compute posterior Dirichlet parameters
# ------------------------------------------------------------

# Dirichlet posterior update:
# posterior_alpha = prior_alpha + observed_counts
#
# Example:
# if prior = [1, 1, 1]
# and counts = [3000, 1400, 1600]
# then posterior = [3001, 1401, 1601]
posterior_alpha = prior_alpha + counts


# ------------------------------------------------------------
# 3. Compute posterior means
# ------------------------------------------------------------

# Posterior mean for each category:
# posterior_alpha_j / sum(posterior_alpha)
#
# This gives Bayesian-smoothed estimated proportion of each contract type.
posterior_means = posterior_alpha / posterior_alpha.sum()

# MLE proportions:
# raw observed category proportions from data
mle_proportions = counts / counts.sum()


# ------------------------------------------------------------
# Self-checks
# ------------------------------------------------------------

# There should be exactly 3 posterior parameters,
# one for each contract category.
assert len(posterior_alpha) == 3

# Total posterior count should equal prior total + observed total count.
assert posterior_alpha.sum() == prior_alpha.sum() + counts.sum()

# Print results
print(f"Counts: {counts}")
print(f"Posterior Dirichlet params: {posterior_alpha}")
print(f"Posterior means: {posterior_means}")
print(f"MLE proportions: {mle_proportions}")
print("✅ Dirichlet posterior computed!")


# ------------------------------------------------------------
# Plot marginal Beta posteriors for each category
# ------------------------------------------------------------
# In a Dirichlet distribution, each category's marginal distribution
# follows a Beta distribution:
#
# θ_j ~ Beta(alpha_j, total_alpha - alpha_j)
#
# This lets us plot uncertainty for each category proportion separately.
# ------------------------------------------------------------

# Create possible proportion values from 0 to 1
theta_r = np.linspace(0.001, 0.999, 500)

# Create 1 row and 3 columns of plots
# One plot for each contract category
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Loop over each subplot, category name, and posterior alpha value
for ax, (cat, a_j) in zip(axes, zip(categories, posterior_alpha)):

    # For category j:
    # a_j = posterior alpha for that category
    #
    # b_j = sum of posterior alphas of all other categories
    # This forms the marginal Beta(a_j, b_j)
    b_j = posterior_alpha.sum() - a_j

    # Compute marginal Beta posterior PDF for this category
    pdf = beta_dist.pdf(theta_r, a_j, b_j)

    # Compute posterior mean for this category
    posterior_mean = a_j / posterior_alpha.sum()

    # Compute raw MLE proportion for this category
    mle_prop = counts[categories.index(cat)] / counts.sum()

    # Plot the marginal posterior distribution
    ax.plot(theta_r, pdf, color="steelblue", lw=2.5, label="Posterior PDF")

    # Draw vertical line for posterior mean
    ax.axvline(
        posterior_mean,
        color="orange",
        linestyle="-.",
        lw=2,
        label=f"Post mean={posterior_mean:.3f}"
    )

    # Draw vertical line for MLE proportion
    ax.axvline(
        mle_prop,
        color="crimson",
        linestyle="--",
        lw=2,
        label=f"MLE={mle_prop:.3f}"
    )

    # Set plot title showing category and marginal Beta parameters
    ax.set_title(f"{cat}\nBeta({a_j:.0f},{b_j:.0f})")

    # X-axis = possible proportion of that contract category
    ax.set_xlabel("Proportion θ")

    # Y-axis = density
    ax.set_ylabel("Density")

    # Show legend
    ax.legend(fontsize=8)


# Main title for all three plots
plt.suptitle(
    "Dirichlet-Multinomial: Marginal Beta Posteriors per Category",
    fontsize=12
)

# Adjust spacing between plots
plt.tight_layout()

# Display final plot
plt.show()
````

````python
# ------------------------------------------------------------
# Add an unseen category: "Biannual"
# ------------------------------------------------------------
# Current categories are:
# 1. Month-to-month
# 2. One year
# 3. Two year
#
# Now we imagine a new category called "Biannual".
# This category has no observations in the dataset.
#
# If we used raw counts only, its count would be 0.
# But Dirichlet prior allows us to give it a small pseudocount.
# Here, we add pseudocount = 1.
# ------------------------------------------------------------


# Add a 4th Dirichlet parameter for the unseen "Biannual" category.
# posterior_alpha already contains posterior counts for the 3 observed categories.
# np.append(...) adds 1.0 as the alpha value for the new unseen category.
new_alpha_4cat = np.append(posterior_alpha, 1.0)


# Create a new category list including the unseen category.
new_categories = categories + ["Biannual"]


# ------------------------------------------------------------
# Compute posterior mean for the unseen category
# ------------------------------------------------------------

# Total posterior mass across all 4 categories
# This is denominator for posterior mean calculation.
total_alpha_4cat = new_alpha_4cat.sum()


# The unseen category is the last category, so its alpha is new_alpha_4cat[-1].
biannual_alpha = new_alpha_4cat[-1]


# Posterior mean for unseen category:
# alpha_j / sum(all alpha)
biannual_mean = biannual_alpha / total_alpha_4cat


# For comparison, compute posterior mean of Month-to-month category.
# Month-to-month has many observations, so its posterior mean will be much larger.
month_to_month_mean = new_alpha_4cat[0] / total_alpha_4cat


# Print results
print("Posterior for unseen 'Biannual' category:")
print(f"New 4-category posterior alpha: {new_alpha_4cat}")
print(f"Biannual posterior mean: {biannual_mean:.6f}")
print(f"Month-to-month posterior mean: {month_to_month_mean:.6f}")


# ------------------------------------------------------------
# Marginal posterior for unseen Biannual category
# ------------------------------------------------------------
# In a Dirichlet distribution, each category's marginal posterior is Beta:
#
# θ_j ~ Beta(alpha_j, total_alpha - alpha_j)
#
# For Biannual:
# alpha = 1 because it has only prior pseudocount
# beta  = total alpha of all other categories
# ------------------------------------------------------------

# Alpha for Biannual category
a_biannual = biannual_alpha

# Beta for Biannual category = all other category mass
b_biannual = total_alpha_4cat - a_biannual


# Possible values of θ from 0 to 1
theta_r = np.linspace(0.001, 0.999, 500)


# Compute marginal Beta posterior PDF for Biannual proportion
biannual_pdf = beta_dist.pdf(theta_r, a_biannual, b_biannual)


# ------------------------------------------------------------
# Plot marginal posterior for unseen Biannual category
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

# Plot posterior PDF
plt.plot(
    theta_r,
    biannual_pdf,
    color="steelblue",
    lw=2.5,
    label="Biannual posterior PDF"
)

# Draw vertical line for posterior mean
plt.axvline(
    biannual_mean,
    color="orange",
    linestyle="-.",
    lw=2,
    label=f"Posterior mean = {biannual_mean:.6f}"
)

# Add title and labels
plt.title(f"Unseen Category: Biannual\nBeta({a_biannual:.0f}, {b_biannual:.0f})")
plt.xlabel("Proportion θ")
plt.ylabel("Density")

# Show legend
plt.legend()

# Adjust layout and display plot
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Interpretation
# ------------------------------------------------------------

print("\nInterpretation:")
print(
    "The unseen 'Biannual' category receives a small non-zero posterior probability "
    "because the Dirichlet prior gives it a pseudocount of 1."
)
print(
    "However, its posterior mean is very small compared to Month-to-month because "
    "there are no actual observations for Biannual, while Month-to-month has many."
)
````
